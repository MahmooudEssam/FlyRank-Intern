### Accessing your Hugging Face Token from Secrets

Once you have added your token with the name `HF_TOKEN` to Colab's Secrets manager, you can retrieve it dynamically in your code without exposing it:

```python
import duckdb
from google.colab import userdata

# Retrieve your HF token from Colab secrets securely
hf_token = userdata.get('hf_token')

con = duckdb.connect()
con.execute(f"CREATE SECRET (TYPE huggingface, TOKEN '{hf_token}')")

rel = "hf://datasets/FlyRank/internship-warehouse"
res = con.sql(f"SELECT * FROM read_parquet('{rel}/fact_content_daily_performance/**/*.parquet') LIMIT 5")
display(res)
```

### Manipulating the Data with Pandas

You can convert the DuckDB query results directly into a Pandas DataFrame using the `.df()` method. This allows you to inspect, filter, group, and modify the data just as you would with a loaded CSV file.

```python
# Convert the DuckDB query result to a pandas DataFrame
df = res.df()

# 1. Preview the first few rows and column types
print("--- Data Overview ---")
display(df[['gsc_impressions', 'gsc_sum_position', 'gsc_avg_position']].head())
```

### Investigating Potential Actions (Page 2 Opportunities)

Let's calculate how many pages have an average position on the second page of search results (between 11 and 20) but still receive high impressions. This represents low-hanging fruit for content optimization.

```python
# Load the full dataset to find specific actionable optimization targets
full_df = con.sql(f"SELECT content_hash_id, gsc_impressions, gsc_avg_position FROM read_parquet('{rel}/fact_content_daily_performance/**/*.parquet')").df()

# Filter for pages on Page 2 of Google (position between 11 and 20)
page_2_opportunities = full_df[(full_df['gsc_avg_position'] > 10) & (full_df['gsc_avg_position'] <= 20)]

# Sort by impressions to find high-opportunity pages
page_2_opportunities = page_2_opportunities.sort_values(by='gsc_impressions', ascending=False)

print(f"Total pages sitting on Page 2: {len(page_2_opportunities)}")
print("Top 5 high-impression pages on Page 2 ripe for optimization:")
display(page_2_opportunities.head(5))
```

## 4. Careful words: what I can and can't claim

### What my work CAN claim:
* **Observed Correlations:** We can observe and highlight which historical content performance metrics (e.g., specific thresholds of impressions and mid-tier rankings) consistently correlate with traffic surges after an optimization action is taken.
* **Directional Value:** We can establish directional indicators for content optimization priority (e.g., identifying high-intent pages sitting on Page 2 that present a lower-effort, higher-probability path to Traffic Growth than starting fresh).
* **Decision-Support:** This model acts as a decision-support filter for editors to prioritize their content pipeline based on historical probability of success rather than intuition alone.

### What my work CANNOT claim:
* **Causal Proof:** We cannot claim that modifying a page *causally* guarantees a rank jump, as search engine results are subject to external forces (competitor updates, Google core updates).
* **Predicting Google:** We will never claim to 'crack' or fully predict the inner workings of Google's ranking algorithms.
